# Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge, Lasso, ElasticNet
```

# **1. Load Dataset**

```python
df = pd.read_csv('diabetes.csv')
df
```

## Convert DataFrame to Numpy arrays

```python
Data = ['Pregnancies', 'BloodPressure', 'BMI', 'Age']

x = df[Data].to_numpy()
y = df['Glucose'].to_numpy()
x.shape, y.shape
```

```python
seed = 0
test_size = 0.3

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=test_size, random_state=seed)
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python

print(f'Number of training samples: {x_train.shape[0]}')
print(f'Number of test samples: {x_test.shape[0]}')
```

# **2. Create a Linear Regression Model**

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

## Train model on training set

```python
model.fit(x_train, y_train)
```

## Evaluate Model on test set

```python
y_pred = model.predict(x_test)
y_pred
```

## Μέτρηση Σφάλματος

Σε αυτή τη περίπτωση της γραμμικής αναδρομής, μετράμε πόσο απέχει η πρόβλεψη από την πραγματική τιμή.

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print(f'Mean Absolute Error (MAE): {round(mae, 2)}')
print(f'Mean Squared Error (MSE): {round(mse, 2)}')
```

Στην ερώτηση ποια μετρική προτιμούμε η σωστή απάντηση έιναι η MAE. Στη προκειμένη περίπτωση εφόσον έχουμε να υπολογίσουμε τις μονάδες γλυκόζης, είναι πιο χρήσιμη η MAE καθώς βρίσκεται στις ίδιες μονάδες με την γλυκόζη.

# **3. Ποιά μεταβλητή είναι πιο σημαντική**

Στα δεδομένα μου, υπάρχουν διαφορετικές κλίμακες στις μεταβλητές μου (π.χ. οι εγκυμοσύνες έιναι πιο μικρός ο αριθμός τους απο την πίεση). Επομένως κάνω όλες τις μεταβλητές μου στην ίδια κλίμακα

## Κανονικοποίηση με Z-Score

```python
scaler = StandardScaler()
scaler.fit(x_train)                      # υπολογίζει μεση τιμή και απόκλιση μόνο από το train

x_train_new = scaler.transform(x_train)
x_test_new = scaler.transform(x_test) # εφαρμόζεται στο test

print('μ (μέση τιμή) κάθε χαρακτηριστικού:', scaler.mean_.round(2))
print('σ (τυπική απόκλιση) κάθε χαρακτηριστικού:', scaler.scale_.round(2))
```

##Γραμμική Παλινδρόμηση στα κανονικοποιημένα δεδομένα

```python
model_new = LinearRegression()
model_new.fit(x_train_new, y_train)

coef_new = pd.Series(model_new.coef_, index=Data)
print('Συντελεστές:')
print(coef_new.round(2))
print()
print(f'MAE στο test: {mean_absolute_error(y_test, model_new.predict(x_test_new).round(2))}')
```

Μετά την κανονικοποίηση, η πιο σημαντική μεταβλητή φαίνεται να είναι η Age (w ≈ 9.21 ) και ακολουθεί η BMI (w ≈ 7.56)

##Ομαλοποίηση

Χρησιμοποιούμε λ = 1

```python
ridge = Ridge(alpha=1)
ridge.fit(x_train_new, y_train)

lasso = Lasso(alpha=1)
lasso.fit(x_train_new, y_train)

elastic = ElasticNet(alpha=1, l1_ratio=0.5)
elastic.fit(x_train_new, y_train)

coefs = pd.DataFrame({
    'Linear Regression': model_new.coef_,
    'Ridge (L2)': ridge.coef_,
    'Lasso (L1)': lasso.coef_,
    'Elastic Net': elastic.coef_,
}, index=Data)
coefs.round(2)
```

Παρατηρήσεις:

- Ridge (L2): σχεδόν ίδιοι συντελεστές με τη Γραμμική Παλινδρόμηση, αφού το μοντέλο έχει μόνο 4 χαρακτηριστικά και δεν κάνει overfitting. Η Age παραμένει πρώτη.
- Lasso (L1):μηδενίζει τα Pregnancies (w = 0) και σχεδόν μηδενίζει τη BloodPressure (w ≈ 0.21), δηλαδή τα κρίνει ασήμαντα. Κρατάει μόνο την Age (≈ 8.04) και το BMI (≈ 6.74).
- Elastic Net: μικραίνει όλα τα βάρη, αλλά η σειρά μένει ίδια, με πρώτη την Age.


##Συμπέρασμα

Η πιο σημαντική ανεξάρτητη μεταβλητή για την πρόβλεψη της γλυκόζης είναι η ηλικία (Age), με δεύτερο το BMI. Μετά την κανονικοποίηση (Z-Score), που φέρνει όλα τα χαρακτηριστικά σε κοινή κλίμακα, η Age έχει το μεγαλύτερο |w|.
Η ομαλοποίηση το επιβεβαιώνει: σε Ridge, Lasso και Elastic Net η Age έχει πάντα το μεγαλύτερο |w|, ενώ η Lasso μηδενίζει τα Pregnancies, δείχνοντας ότι δεν συμβάλλουν στην πρόβλεψη.
